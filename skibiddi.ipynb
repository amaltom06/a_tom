```python
import os
import gradio as gr
from langchain_groq import ChatGroq
from langchain_core.output_parsers.string import StrOutputParser
from langchain_core.prompts import ChatPromptTemplate

# Set up the API key
os.environ["GROQ_API_KEY"] = "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"

# Define system prompt and model
system_prompt = """
You are Skibiddi Bot, a super chill and entertaining assistant.
Use fun Gen Z slang and emojis to respond to the user's questions.
When the conversation ends, say "Catch ya later, fam! ✌️".
"""
prompt = ChatPromptTemplate.from_messages(
    [("system", system_prompt), ("human", "{input}")]
)
llm = ChatGroq(temperature=0.7, model="llama3-8b-8192")  # Main model for general queries
chain = prompt | llm | StrOutputParser()

# Dictionary of queries and responses about Amal Tom Parakkaden
amal_queries = {
    "who is amal tom": """
**Amal Tom Parakkaden**
🌟 I'm an MCA student passionate about cybersecurity, blending software development skills with hands-on experience in vulnerability assessments and penetration testing.
""",
    "educational background": """
🎓 **Educational Background:**
- **MCA** | 2023-2025 | Muthoot Institute of Technology and Science
- **Certified IT Infrastructure and Cyber SOC Analyst** | 2022-2023 | RedTeam Hacker Academy
- **BCA** | 2019-2022 | Mar Gregorios College of Arts and Science
""",
    "professional experience": """
💼 **Professional Experience:**
- **Internship at InternPe** (Python) | Developed and debugged Python code.
- **Internship at Plasmid** (Cybersecurity) | Conducted penetration testing and vulnerability assessments.
""",
    "skills": """
🛠️ **Skills:**
HTML, CSS, Python, Kali Linux, Penetration Testing, and more!
""",
    "contact information": """
📧 **Contact:** amaltomparakkaden@gmail.com
📞 **Phone:** +91-8137819187
""",
    "career goals": """
🚀 **Career Goals:**
My short-term goal is to secure a position in cybersecurity where I can apply my skills in vulnerability assessment and penetration testing.
""",
    "philosophy": """
🔍 **Philosophy:**
I believe in continuous learning and staying updated with the latest trends in cybersecurity.
"""
}

# Define a function to generate responses
def get_skibiddi_response(user_question):
    user_question = user_question.lower()

    # Check for specific queries about Amal
    for query, response in amal_queries.items():
        if query in user_question:
            return response

    # If no specific query, generate a normal response using the Llama model
    response = chain.invoke({"input": user_question})
    return f"{response} 😎✨"

# Create a colorful Gradio interface
interface = gr.Interface(
    fn=get_skibiddi_response,
    inputs=gr.Textbox(placeholder="What's on your mind? 🧠💭", label="Ask Skibiddi Bot"),
    outputs=gr.Textbox(label="Skibiddi Bot's Response"),
    title="Skibiddi Bot 🤖✨",
    description="Chat with Skibiddi Bot, your ultimate Gen Z assistant! Hit me up with your questions! 🥳",
    theme="compact",
    css="""
        body { background-color: #ffccff; font-family: 'Comic Sans MS', cursive; }
        .input_text { background-color: #ffeb3b; border-radius: 10px; }
        .output_text { background-color: #e0f7fa; border-radius: 10px; }
        .title { color: #d5006d; text-shadow: 1px 1px 2px #fff; }
        .description { color: #6200ea; }
        .input_text:focus { border: 2px solid #d5006d; }
        .output_text { padding: 10px; font-size: 1.2em; }
    """
)

# Launch the interface
interface.launch()
```
